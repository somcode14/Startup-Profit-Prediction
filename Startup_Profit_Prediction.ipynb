# Exploratory Data Analysis (EDA)

This notebook performs basic exploratory data analysis on a given dataset using Python and Pandas.


```python
## 1. Importing Required Libraries
import pandas as pd
import numpy as np
import matploistlib.pyplot as plt
```

```python
## 2. Loading the Dataset
df=pd.read_csv("/content/50_Startups.csv")
```

```python
## 3. Displaying First Few Rows
df.head(2)
```

```python
## 4. Checking Dataset Shape
df.shape
```

```python
## 5. Dataset Information
df.info()
```

```python
## 6. Checking Missing Values
df.isnull().sum()
```

```python
## 7. Statistical Summary
df.describe()
```

```python
## 8. Checking Frequency of Each State
df["State"].value_counts()
```

```python
## 9. Creating a State Encoding Dictionary

state_dict = {
    "California": 0,
    "New York": 1,
    "Florida": 2
}
```

```python
## 10. Converting State Names into Numerical Values
df["State_num"] = df["State"].map(state_dict)
df["State_num"]
```

```python
## This step separates the independent variables (X) by dropping Profit and
## State columns and sets Profit as the dependent variable (y).
x = df.drop(["Profit", "State"], axis=1)
y = df["Profit"]
```

```python
## Checking Shape of X and Y
print("shape of x", x.shape)
print("shape of y", y.shape)
```

```python
## Importing Train-Test Split Function
from sklearn.model_selection import train_test_split
```

```python
## Splitting Data into Training and Testing Sets
x_train, x_test, y_train, y_test=train_test_split(x,y,test_size=0.2, random_state=0)
```

```python
## Checking Shape of Training and Testing Dataprint("shape of x_train: ",x_train.shape)
print("shape of x_test: ",x_test.shape)
```

```python
## Checking Shape of Training and Testing Labels
print("shape of y_train: ",y_train.shape)
print("shape of y_test: ",y_test.shape)
```

```python
## Importing Linear Regression Model
from sklearn.linear_model import LinearRegression
```

```python
lr = LinearRegression()
```

```python
## Training the Linear Regression Model
lr.fit(x_train, y_train)
```

```python
## Importing Model Evaluation Metrics
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
```

```python
## Displaying Predicted Values
y_pred=lr.predict(x_test)
y_pred
```

```python
#MAE
print("MAE value: " , mean_absolute_error(y_test,y_pred))
```

```python
#MSE
print("MSE value: ",  mean_squared_error(y_test,y_pred))
```

```python
#RMSE
print("RMSE value: ",  np.sqrt(mean_squared_error(y_test,y_pred)))
```

```python
## Calculating R² Score
print("R2 Score: ",r2_score(y_test,y_pred))
```
